## 1. Executar somente quando uma condição for verdadeira

O bloco que vem depois de `if` é executado quando a condição é verdadeira. Os dois-pontos iniciam o bloco; a indentação mostra quais linhas pertencem a ele.

Uma linha sem essa indentação volta ao fluxo externo.

**Preveja:** o que muda se `temperatura` for `80.0`? A última mensagem ainda será exibida?


```python
temperatura = 120.0
limite = 100.0

if temperatura > limite:
    print("A temperatura ultrapassou o limite.")
    print(f"Excesso: {temperatura - limite:.1f} °C")

print("A verificação terminou.")
```

## 2. Escolher entre alternativas

`elif` testa outra condição quando as anteriores foram falsas. `else` cobre o caso restante. Em uma sequência `if`/`elif`/`else`, no máximo um desses blocos será executado.

As faixas abaixo são uma classificação definida para o exemplo, não uma norma de medição.

**Altere:** teste `-2.0`, `0.0`, `50.0`, `100.0` e `120.0`. Observe em qual faixa entram os limites.


```python
temperatura = 25.0

if temperatura < 0.0:
    classificacao = "abaixo da faixa"
elif temperatura <= 100.0:
    classificacao = "dentro da faixa"
else:
    classificacao = "acima da faixa"

print(f"{temperatura:.1f} °C: {classificacao}")
```

## 3. Combinar condições sem perder a clareza

Podemos usar uma comparação encadeada, como `0.0 <= valor <= 1.0`, ou combinar condições com `and` e `or`. Nomes intermediários deixam a regra mais fácil de ler.

**Altere:** faça primeiro a leitura ficar fora da faixa e depois marque o sensor como indisponível. A mensagem permite identificar o motivo?


```python
leitura = 0.75
sensor_disponivel = True
leitura_na_faixa = 0.0 <= leitura <= 1.0

if not sensor_disponivel:
    print("O sensor está indisponível.")
elif not leitura_na_faixa:
    print("A leitura está fora da faixa.")
else:
    print("A leitura está disponível e dentro da faixa.")
```

## 4. Percorrer uma lista com `for`

Um laço `for` executa seu bloco uma vez para cada item da sequência. O nome `medicao` recebe um valor diferente em cada passagem.

Para acumular uma soma, precisamos definir seu valor inicial antes do laço. A operação `soma += medicao` é uma forma curta de escrever `soma = soma + medicao`.

**Preveja:** qual será a soma? Depois acrescente uma medição e execute a célula inteira novamente.


```python
medicoes = [20.0, 21.0, 22.0]
soma_medicoes = 0.0

for medicao in medicoes:
    soma_medicoes += medicao
    print(f"Medição: {medicao:.1f}; soma parcial: {soma_medicoes:.1f}")

media_medicoes = soma_medicoes / len(medicoes)
print(f"Média: {media_medicoes:.1f}")
```

## 5. Contar com `range`

`range(inicio, fim, passo)` produz uma sequência de inteiros para percorrer. O limite final não está incluído. Com um argumento, o início é zero; o passo padrão é um.

Usamos `list` abaixo apenas para visualizar os valores. Um `for` pode percorrer o `range` diretamente.

**Preveja:** quantas vezes o laço será executado? Depois faça a contagem começar em `1` e terminar em `4`.


```python
print(list(range(4)))
print(list(range(1, 5)))
print(list(range(0, 7, 2)))

for indice in range(4):
    print(f"Repetição de índice {indice}")
```

## 6. Percorrer valores e seus índices com `enumerate`

Quando precisamos tanto do índice quanto do valor, `enumerate` fornece os dois. Por padrão, a contagem começa em zero.

O índice pode identificar a posição de uma medição ou de um nó. Evite chamar todas as variáveis de `x`: um nome como `indice` distingue a posição na lista do valor guardado nela.

**Altere:** acrescente uma posição à lista e observe a numeração.


```python
posicoes = [0.0, 0.25, 0.50, 0.75, 1.0]

for indice, posicao in enumerate(posicoes):
    print(f"Nó {indice}: posição {posicao:.2f}")
```

## 7. Percorrer duas sequências com `zip`

`zip` associa o primeiro item de uma sequência ao primeiro da outra, depois os segundos, e assim por diante. Por padrão, ele para quando a sequência mais curta termina.

Por isso, confira se os comprimentos são compatíveis quando cada posição deve ter uma medição. Não use `zip` para esconder uma diferença de tamanho nos dados.

**Altere:** retire uma temperatura. A condição abaixo evita apresentar pares incompletos.


```python
posicoes = [0.0, 0.5, 1.0]
temperaturas = [20.0, 25.0, 22.0]

if len(posicoes) != len(temperaturas):
    print("É necessário ter uma temperatura para cada posição.")
else:
    for posicao, temperatura in zip(posicoes, temperaturas):
        print(f"x = {posicao:.1f}: temperatura = {temperatura:.1f} °C")
```

## 8. Repetir enquanto uma condição for verdadeira

Um laço `while` testa a condição antes de cada passagem. Se ela começar falsa, o bloco não é executado.

O exemplo reduz uma quantidade pela metade até que ela fique menor ou igual a uma tolerância, ou até atingir um limite de repetições. Atualizar `quantidade` e `numero_repeticoes` é essencial para o processo terminar.

**Preveja:** quantas divisões serão necessárias? Depois reduza `maximo_repeticoes` para `2` e observe que atingir o limite não significa atingir a tolerância.


```python
quantidade = 1.0
tolerancia = 0.1
numero_repeticoes = 0
maximo_repeticoes = 10

while quantidade > tolerancia and numero_repeticoes < maximo_repeticoes:
    quantidade = quantidade / 2.0
    numero_repeticoes += 1

print(f"Quantidade final: {quantidade}")
print(f"Número de repetições: {numero_repeticoes}")
print(f"Atingiu a tolerância? {quantidade <= tolerancia}")
```

## 9. Interromper um laço com `break`

`break` encerra o laço mais interno em que aparece. Podemos usá-lo quando uma busca já encontrou o que procurava.

O valor inicial `None` permite distinguir “nenhum item encontrado” de um resultado numérico.

**Altere:** faça o limite ser maior que todos os valores da lista. Qual resultado será apresentado?


```python
leituras = [0.2, 0.4, 0.9, 1.1, 0.3]
limite = 0.8
primeira_leitura_acima = None

for leitura in leituras:
    if leitura > limite:
        primeira_leitura_acima = leitura
        break

print("Primeira leitura acima do limite:", primeira_leitura_acima)
```

## 10. Pular uma passagem com `continue`

`continue` pula o restante da passagem atual e segue para a próxima. Diferentemente de `break`, ele não encerra o laço inteiro.

Neste exemplo, `None` identifica uma medição ausente. O valor `0.0` é uma medição válida e deve participar da soma. A verificação de quantidade evita dividir por zero quando nenhuma medição está disponível.

**Altere:** use `[None, None]` e confira a mensagem.


```python
medicoes = [20.0, None, 0.0, 22.0]
soma_medicoes = 0.0
quantidade_validas = 0

for medicao in medicoes:
    if medicao is None:
        continue
    soma_medicoes += medicao
    quantidade_validas += 1

if quantidade_validas == 0:
    print("Não há medições disponíveis.")
else:
    print(f"Média das medições disponíveis: {soma_medicoes / quantidade_validas}")
```

## 11. Um laço dentro de outro

Laços aninhados permitem percorrer combinações de índices. O laço externo abaixo escolhe a linha; o interno escolhe a coluna.

Estamos apenas construindo uma tabela de números com listas. Não há aqui uma implementação de elementos finitos nem uma biblioteca de matrizes.

**Preveja:** quantos valores serão criados? Observe que `linha = []` precisa ser executado para cada nova linha.


```python
numero_linhas = 2
numero_colunas = 3
tabela = []

for indice_linha in range(numero_linhas):
    linha = []
    for indice_coluna in range(numero_colunas):
        valor = 10 * indice_linha + indice_coluna
        linha.append(valor)
    tabela.append(linha)

print(tabela)
print("Valor da linha 1, coluna 2:", tabela[1][2])
```

## 12. Construir uma lista: primeiro com um laço

Uma *list comprehension* é uma forma de construir uma lista a partir de uma sequência. Antes de usá-la, vamos escrever o laço equivalente.

Nos dois casos abaixo, a operação é a mesma: elevar cada valor ao quadrado e reunir os resultados em outra lista. A comprehension tem a forma `[expressao for item in sequencia]`.

**Confira:** as duas listas são iguais? Depois calcule o dobro de cada valor usando as duas formas.


```python
valores = [0.0, 0.25, 0.5, 1.0]

quadrados_com_laco = []
for valor in valores:
    quadrados_com_laco.append(valor**2)

quadrados_com_compreensao = [valor**2 for valor in valores]

print(quadrados_com_laco)
print(quadrados_com_compreensao)
print(quadrados_com_laco == quadrados_com_compreensao)
```

## 13. Filtrar uma lista: comparar as duas formas

Uma comprehension também pode incluir uma condição: `[expressao for item in sequencia if condicao]`.

Ela é útil quando a operação cabe em uma linha legível. Para várias decisões, mensagens ou etapas de cálculo, prefira um laço explícito.

**Preveja:** o zero aparece na lista resultante? Depois mude a condição para selecionar somente valores positivos.


```python
valores = [-2.0, 0.0, 1.5, -0.5, 3.0]

nao_negativos_com_laco = []
for valor in valores:
    if valor >= 0.0:
        nao_negativos_com_laco.append(valor)

nao_negativos_com_compreensao = [valor for valor in valores if valor >= 0.0]

print(nao_negativos_com_laco)
print(nao_negativos_com_compreensao)
```

## 14. Preservar os dados de entrada

É mais fácil conferir um cálculo quando distinguimos os dados recebidos dos resultados produzidos. Neste exemplo, a lista `medicoes` permanece igual e a correção é guardada em `medicoes_corrigidas`.

Evite acrescentar ou remover itens da mesma lista que está sendo percorrida: isso muda os dados durante a repetição e pode produzir resultados difíceis de prever.

**Altere:** use outra correção e confira tanto a lista original quanto a nova.


```python
medicoes = [20.0, 21.5, 19.0]
correcao = 0.5
medicoes_corrigidas = []

for medicao in medicoes:
    medicoes_corrigidas.append(medicao + correcao)

print("Dados originais:", medicoes)
print("Dados corrigidos:", medicoes_corrigidas)
```
