## 3. Um primeiro cálculo

Vamos dividir um intervalo de comprimento 2 em quatro partes iguais. Antes de executar, calcule mentalmente o comprimento de cada parte.

````python
comprimento = 2.0
numero_partes = 4
tamanho_parte = comprimento / numero_partes
print(tamanho_parte)
````

O resultado é `0.5`. Os nomes à esquerda de `=` guardam referências aos valores usados no cálculo. Os detalhes da atribuição serão estudados no caderno 01.

**Experimente:** mantenha o comprimento e use oito partes. Qual resultado você prevê? Modifique, execute e compare.

````python
comprimento_exercicio = 2.0
numero_partes_exercicio = 4  # Altere este valor para experimentar.
print(comprimento_exercicio / numero_partes_exercicio)
````

## 5. Salvar não é executar

Salvar o arquivo preserva o texto, o código e as saídas registradas no notebook. Não salva todas as variáveis que existem na memória do kernel. Ao reabrir o material, execute as células para reconstruir o cálculo.

Use **Ctrl+S** (ou **Cmd+S** no macOS) para salvar. Na aula de Git veremos que salvar um arquivo também é diferente de registrar uma versão dele no histórico.
