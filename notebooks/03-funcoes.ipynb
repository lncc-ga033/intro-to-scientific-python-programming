## 1. Definir uma função e chamá-la

`def` inicia a definição. Os parâmetros `largura` e `altura` dão nome às entradas dentro da função. `return` devolve o resultado para quem a chamou.

Definir a função não executa seu cálculo. O cálculo acontece quando escrevemos uma chamada, como `area_retangulo(2.0, 3.0)`.

**Preveja:** qual valor será guardado em `area`? Depois faça outra chamada com medidas diferentes, sem copiar o cálculo.


```python
def area_retangulo(largura, altura):
    area = largura * altura
    return area


area = area_retangulo(2.0, 3.0)
print(f"Área: {area}")
print(f"Outra área: {area_retangulo(4.0, 1.5)}")
```

## 2. Argumentos por posição ou por nome

Os argumentos são os valores fornecidos na chamada. Podemos associá-los pela ordem dos parâmetros ou escrever explicitamente seus nomes.

Argumentos nomeados ajudam quando dois valores do mesmo tipo têm papéis diferentes. O nome usado na chamada precisa coincidir com o parâmetro da função.

**Altere:** experimente outro par de medidas nas duas chamadas e confira se os resultados coincidem.


```python
area_por_posicao = area_retangulo(2.0, 3.0)
area_por_nome = area_retangulo(altura=3.0, largura=2.0)

print(area_por_posicao)
print(area_por_nome)
```

## 3. Devolver um resultado é diferente de imprimi-lo

`print` mostra uma mensagem. `return` permite usar o resultado em outro cálculo. Uma função sem `return` explícito devolve `None`.

Em computação científica, é útil separar a função que calcula da instrução que apresenta o resultado. Assim, o mesmo cálculo pode alimentar uma tabela, um gráfico ou uma verificação.

**Preveja:** o que será guardado em `retorno_impressao`?


```python
def calcular_dobro(valor):
    return 2.0 * valor


def imprimir_dobro(valor):
    print(f"Dobro: {2.0 * valor}")


dobro = calcular_dobro(3.0)
retorno_impressao = imprimir_dobro(3.0)

print(f"Resultado que posso reutilizar: {dobro}")
print(f"Retorno da função que só imprime: {retorno_impressao}")
print(f"Resultado de outro cálculo: {dobro + 1.0}")
```

## 4. Argumentos com valores padrão

Um parâmetro pode receber um valor padrão, usado quando o argumento é omitido. Os parâmetros obrigatórios aparecem antes dos parâmetros com padrão.

O padrão deve ter um significado claro. Na função abaixo, não informar `fator` significa usar `1.0`, mantendo o valor recebido.

**Altere:** use `fator=0.5`. O que acontece com o resultado?


```python
def multiplicar_por_fator(valor, fator=1.0):
    return fator * valor


print(multiplicar_por_fator(8.0))
print(multiplicar_por_fator(8.0, 2.0))
print(multiplicar_por_fator(8.0, fator=0.5))
```

## 5. Retornar mais de uma informação

Podemos devolver uma tupla e desempacotar seus itens na chamada. A função abaixo calcula o ponto médio e o comprimento de um intervalo.

Essa forma é adequada para poucos resultados, com uma ordem fácil de explicar. Mais adiante veremos alternativas para agrupar dados quando o programa crescer.

**Preveja:** qual variável receberá o ponto médio? Trocar a ordem no desempacotamento muda o significado dos nomes.


```python
def descrever_intervalo(inicio, fim):
    ponto_medio = (inicio + fim) / 2.0
    comprimento = fim - inicio
    return ponto_medio, comprimento


ponto_medio, comprimento = descrever_intervalo(1.0, 3.0)
print(f"Ponto médio: {ponto_medio}")
print(f"Comprimento: {comprimento}")
```

## 6. Escopo: um nome dentro e outro fora da função

Os parâmetros e os nomes atribuídos dentro de uma função são locais a ela. A atribuição a `fator` no corpo abaixo não altera o nome `fator` definido fora.

Para cálculos reutilizáveis, prefira receber os dados necessários como argumentos. Depender de valores globais torna mais difícil saber o que precisa estar definido antes da chamada.

**Observe:** esta explicação trata de atribuições a nomes. Uma função que recebe uma lista ainda pode modificar essa lista; veremos abaixo uma função que preserva seus dados de entrada.


```python
fator = 10.0


def aplicar_fator_local(valor):
    fator = 2.0
    return fator * valor


resultado = aplicar_fator_local(3.0)
print(f"Resultado da função: {resultado}")
print(f"Fator definido fora: {fator}")
```

## 7. Documentar o significado da função

Uma *docstring* é o texto entre aspas triplas logo após a definição. Ela pode explicar o cálculo, as entradas e o resultado.

Anotações como `inicio: float` e `-> float` registram os tipos esperados. Elas ajudam a leitura, mas não convertem nem validam os argumentos automaticamente.

**Altere:** melhore a descrição se achar que falta informação. Um comentário deve explicar o significado ou a decisão, não apenas repetir a operação.


```python
def calcular_comprimento(inicio: float, fim: float) -> float:
    """Retorna fim - inicio para um intervalo com extremos na mesma unidade."""
    return fim - inicio


print(calcular_comprimento(0.0, 1.0))
print(calcular_comprimento.__doc__)
```

## 8. Validar entradas com `ValueError`

Uma função pode recusar dados que não fazem sentido para seu cálculo. `raise ValueError(...)` interrompe a chamada e informa qual regra foi violada.

Na função abaixo, estamos adotando a convenção de que um intervalo válido tem `fim > inicio`. Essa escolha faz parte do contrato da função.

**Observe:** a chamada desta célula é válida. Na próxima, veremos uma chamada inválida com o erro capturado, de modo que a execução do caderno possa continuar.


```python
def comprimento_positivo(inicio, fim):
    """Retorna o comprimento de um intervalo orientado da esquerda para a direita."""
    if fim <= inicio:
        raise ValueError("O extremo final deve ser maior que o inicial.")
    return fim - inicio


print(comprimento_positivo(0.0, 1.0))
```

## 9. Ler um erro e identificar sua origem

`try` executa um bloco; `except ValueError` trata esse tipo específico de erro. Capturamos o erro esperado apenas para poder estudá-lo e continuar o caderno.

Leia primeiro o tipo (`ValueError`) e a mensagem. Depois identifique a chamada e a condição que a função rejeitou. Não ignore uma exceção sem entender sua causa.

**Altere:** use extremos válidos. O bloco `else` abaixo é executado quando a chamada não produz esse erro.


```python
try:
    comprimento = comprimento_positivo(1.0, 0.0)
except ValueError as erro:
    print(f"Erro capturado: {type(erro).__name__}")
    print(f"Mensagem: {erro}")
else:
    print(f"Comprimento calculado: {comprimento}")

print("A execução continuou após a verificação.")
```

## 10. Um argumento opcional que usa `None`

Podemos usar `None` para indicar que um argumento opcional não foi fornecido. A função abaixo devolve todos os valores, ou somente os primeiros, sempre em outra lista.

Evite usar uma lista mutável como valor padrão sem compreender suas consequências: esse objeto seria compartilhado entre chamadas. Aqui o padrão é `None`, e a cópia é criada durante a execução.

**Preveja:** as alterações em `selecionados` mudarão `medicoes`?


```python
def selecionar_primeiros(valores, quantidade=None):
    """Retorna uma cópia de todos os valores ou dos primeiros itens."""
    if quantidade is None:
        return valores.copy()
    return valores[:quantidade]


medicoes = [20.0, 21.0, 22.0]
selecionados = selecionar_primeiros(medicoes, quantidade=2)
selecionados[0] = -1.0

print("Seleção alterada:", selecionados)
print("Entrada preservada:", medicoes)
print("Sem quantidade informada:", selecionar_primeiros(medicoes))
```

## 11. Comparar resultados numéricos com tolerância

`math` é um módulo da biblioteca padrão. A função `math.isclose` compara dois escalares usando uma tolerância relativa (`rel_tol`) e uma absoluta (`abs_tol`).

Para valores finitos, sua regra é verificar se

\[
|a-b|\le\max\bigl(\text{rel\_tol}\,\max(|a|,|b|),\ \text{abs\_tol}\bigr).
\]

A tolerância absoluta é especialmente útil perto de zero. Os valores abaixo servem para estes exemplos curtos; não são uma escolha universal para qualquer simulação.


```python
import math

soma = 0.1 + 0.2

print(soma == 0.3)
print(math.isclose(soma, 0.3, rel_tol=1e-12, abs_tol=1e-12))
print(math.isclose(1e-14, 0.0, rel_tol=1e-12, abs_tol=1e-12))
print(math.isclose(1e-6, 0.0, rel_tol=1e-12, abs_tol=1e-12))
```

## 12. Verificações pequenas e explícitas

`assert` verifica uma condição durante a execução normal do Python. Se ela for falsa, ocorre `AssertionError`. Aqui usamos resultados conhecidos para conferir funções curtas.

`assert` é útil para esses exercícios e verificações de desenvolvimento, mas não substitui a validação com `raise` de dados recebidos por uma função: verificações `assert` podem ser desativadas ao executar Python com otimizações.

**Leia:** cada verificação deve expressar uma propriedade ou resultado que conseguimos justificar independentemente do código.


```python
assert math.isclose(
    area_retangulo(2.0, 3.0),
    6.0,
    rel_tol=1e-12,
    abs_tol=1e-12,
)
assert math.isclose(
    comprimento_positivo(1.2, 3.2),
    2.0,
    rel_tol=1e-12,
    abs_tol=1e-12,
)

print("As duas verificações passaram.")
```

## 13. Funções para uma fonte e uma solução de referência

Funções também representam expressões matemáticas. No intervalo \(0\le x\le1\), considere a fonte constante \(f(x)=2\) e a função \(u(x)=x(1-x)\).

Essas expressões serão úteis no estudo de Poisson. Aqui vamos apenas defini-las e avaliá-las; ainda não estamos construindo um método para resolver a equação.

`fonte` recebe `posicao`, embora o resultado seja constante. Isso mantém a mesma forma de chamada que usaremos para fontes variáveis.

**Preveja:** qual é o valor de `solucao_referencia` nos extremos e no ponto médio?


```python
def fonte(posicao):
    """Retorna o valor da fonte constante f(x) = 2."""
    return 2.0


def solucao_referencia(posicao):
    """Avalia u(x) = x(1 - x), usada como referência em [0, 1]."""
    return posicao * (1.0 - posicao)


for posicao in [0.0, 0.5, 1.0]:
    print(
        f"x = {posicao:.1f}: "
        f"fonte = {fonte(posicao):.1f}; "
        f"referência = {solucao_referencia(posicao):.2f}"
    )
```

## 14. Passar uma função como argumento

Uma função pode receber outra função. A operação `avaliar_em_pontos` abaixo recebe a regra de cálculo e os pontos em que essa regra deve ser avaliada.

Na chamada, usamos `fonte`, sem parênteses, para passar a função. Escrever `fonte(0.0)` calcularia primeiro um número e passaria esse número.

**Altere:** acrescente pontos à lista. A mesma função de avaliação serve para as duas expressões matemáticas.


```python
def avaliar_em_pontos(funcao, pontos):
    """Retorna os valores de uma função escalar nos pontos informados."""
    return [funcao(ponto) for ponto in pontos]


pontos = [0.0, 0.25, 0.5, 0.75, 1.0]
valores_fonte = avaliar_em_pontos(fonte, pontos)
valores_referencia = avaliar_em_pontos(solucao_referencia, pontos)

print("Fonte:", valores_fonte)
print("Referência:", valores_referencia)
```

## 15. Separar cálculo e apresentação

Uma função pequena deve ter uma responsabilidade que conseguimos descrever. `calcular_erros_absolutos` recebe duas sequências de mesmo tamanho e devolve as diferenças em valor absoluto.

A apresentação fica fora da função. Assim podemos mudar o formato da mensagem sem alterar o cálculo.

Os valores aproximados abaixo são dados de exemplo. Esta comparação usa apenas os pontos informados; ela não mede o erro de uma função em todo um intervalo.


```python
def calcular_erros_absolutos(valores, referencias):
    """Retorna os erros absolutos entre duas sequências de mesmo tamanho."""
    if len(valores) != len(referencias):
        raise ValueError("As sequências devem ter o mesmo número de itens.")
    return [abs(valor - referencia) for valor, referencia in zip(valores, referencias)]


pontos = [0.0, 0.25, 0.5, 0.75, 1.0]
referencias = avaliar_em_pontos(solucao_referencia, pontos)
aproximacoes = [0.0, 0.18, 0.24, 0.18, 0.0]
erros = calcular_erros_absolutos(aproximacoes, referencias)

print("Erros absolutos:", erros)
print(f"Maior erro nos pontos avaliados: {max(erros):.3f}")
```
