## 1. Python como calculadora

Os operadores `+`, `-`, `*` e `/` fazem as operações aritméticas usuais. A potência é escrita com `**`. Os parênteses tornam a ordem do cálculo explícita.

`//` calcula o quociente com arredondamento para baixo; `%` calcula o resto. Para números inteiros positivos, podemos usá-los para dividir uma quantidade em grupos.

**Preveja:** `7 / 2`, `7 // 2` e `7 % 2` produzem o mesmo resultado? Depois altere o número `7`.


```python
print(2 + 3 * 4)
print((2 + 3) * 4)
print(2**3)
print(7 / 2)
print(7 // 2)
print(7 % 2)
```

## 2. Nomes, valores e tipos

O sinal `=` atribui um valor a um nome. Nomes como `temperatura` e `numero_medicoes` ajudam a identificar o que está sendo calculado. No código deste curso, usamos nomes em português sem acentos e separamos as palavras com `_`.

`int` representa inteiros; `float`, números de ponto flutuante; `bool`, valores lógicos; `str`, texto. `type` informa o tipo de um objeto.

**Altere:** troque `20.0` por `20` e observe o tipo de `temperatura`. Os dois valores podem participar de cálculos, mas não têm o mesmo tipo.


```python
temperatura = 20.0
numero_medicoes = 4
experimento_ativo = True
nome_material = "aluminio"

print(temperatura, type(temperatura))
print(numero_medicoes, type(numero_medicoes))
print(experimento_ativo, type(experimento_ativo))
print(nome_material, type(nome_material))
```

## 3. Comparações e valores lógicos

Uma comparação produz `True` ou `False`. Para testar igualdade usamos `==`; `=` continua sendo atribuição. Também existem `!=`, `<`, `<=`, `>` e `>=`.

Podemos combinar condições com `and` (e), `or` (ou) e `not` (negação). Essas condições serão usadas no próximo caderno para escolher o que executar.

**Preveja:** quais resultados mudarão se `temperatura` passar de `25.0` para `120.0`?


```python
temperatura = 25.0
limite_inferior = 0.0
limite_superior = 100.0

print(temperatura == 25.0)
print(temperatura != 25.0)
print(temperatura >= limite_inferior and temperatura <= limite_superior)
print(temperatura < limite_inferior or temperatura > limite_superior)
print(not (temperatura > limite_superior))
```

## 4. Texto e apresentação de resultados

Textos ficam entre aspas. Podemos juntar textos com `+`, mas uma *f-string* costuma ser mais clara quando precisamos incluir valores: basta colocar `f` antes das aspas e escrever os nomes entre chaves.

Em `{temperatura:.1f}`, a parte `:.1f` pede uma casa decimal na apresentação. Isso muda a mensagem exibida, não o valor guardado em `temperatura`.

**Altere:** apresente três casas decimais e acrescente o nome de outro material.


```python
material = "Cobre"
temperatura = 23.4567

print("Material: " + material)
print(material.lower())
print(f"A temperatura do {material.lower()} é {temperatura:.1f} °C.")
print(f"O valor usado nos cálculos continua sendo {temperatura}.")
```

## 5. `None`: um valor ainda não informado

`None` representa ausência de um valor. Não significa zero, texto vazio nem `False`. Por exemplo, uma medição ainda não realizada pode ser representada por `None`; uma medição igual a zero já contém um resultado.

Para verificar se um objeto é `None`, usamos `is None`. Não fazemos operações aritméticas com uma medição enquanto ela estiver ausente.

**Altere:** atribua `0.0` a `medicao` e execute novamente.


```python
medicao = None
medicao_zero = 0.0

print(medicao)
print(type(medicao))
print(medicao is None)
print(medicao_zero is None)
```

## 6. Converter texto em número

Um texto como `"12.5"` não é um número. Podemos convertê-lo com `float`. Para texto que representa um inteiro, usamos `int`.

Neste caderno os dados estão escritos nas células, para que a execução completa não precise esperar por uma entrada no teclado. Mais adiante poderemos ler arquivos.

**Observe:** Python usa ponto como separador decimal em literais numéricos e em conversões como esta. A conversão de um texto que não representa um número produz um erro; veremos como ler mensagens de erro no caderno 03.


```python
texto_temperatura = "12.5"
texto_repeticoes = "4"

temperatura = float(texto_temperatura)
repeticoes = int(texto_repeticoes)

print(temperatura + 2.0)
print(repeticoes * 3)
print(str(repeticoes) + " repetições")
```

## 7. Listas: vários valores em uma sequência

Uma lista reúne objetos em uma ordem. Os colchetes delimitam a lista e as vírgulas separam seus itens. `len` informa quantos itens ela contém.

Para um cálculo numérico, é comum manter na mesma lista valores com o mesmo significado. A lista abaixo contém posições ao longo de um intervalo. Uma lista de Python ainda não é um array do NumPy.

**Preveja:** quantas posições estão representadas? Quantos intervalos existem entre elas?


```python
posicoes = [0.0, 0.25, 0.50, 0.75, 1.0]

print(posicoes)
print(f"Número de posições: {len(posicoes)}")
print(f"Número de intervalos consecutivos: {len(posicoes) - 1}")
```

## 8. Indexação: acessar um item

Os índices começam em zero. Em uma lista com cinco itens, os índices não negativos vão de `0` até `4`. Índices negativos contam a partir do fim: `-1` acessa o último item.

Não confunda a posição de um item na lista com o valor numérico guardado nele. Um índice fora dos limites causa `IndexError`.

**Altere:** acesse o segundo e o penúltimo item da lista.


```python
posicoes = [0.0, 0.25, 0.50, 0.75, 1.0]

print(f"Primeiro item: {posicoes[0]}")
print(f"Item de índice 2: {posicoes[2]}")
print(f"Último item: {posicoes[-1]}")
```

## 9. Fatias: acessar parte da sequência

A escrita `lista[inicio:fim]` seleciona do índice inicial até antes do índice final. O limite final não entra na seleção. Podemos omitir um dos limites ou acrescentar um passo: `lista[inicio:fim:passo]`.

Uma fatia pode resultar em uma lista vazia. Isso é diferente de tentar acessar um único índice inexistente.

**Preveja:** quais itens aparecerão em `posicoes[1:4]` e em `posicoes[::2]`?


```python
posicoes = [0.0, 0.25, 0.50, 0.75, 1.0]

print(posicoes[1:4])
print(posicoes[:3])
print(posicoes[3:])
print(posicoes[::2])
print(posicoes[::-1])
```

## 10. Listas podem ser modificadas

Uma lista é mutável: podemos substituir um item, acrescentar um novo com `append` e remover o último com `pop`. Essas operações modificam a própria lista.

**Preveja:** a quantidade de itens muda quando substituímos `medicoes[1]`? E quando usamos `append`?


```python
medicoes = [20.0, 21.0, 19.5]
medicoes[1] = 21.2
print("Depois da substituição:", medicoes)

medicoes.append(20.4)
print("Depois de acrescentar:", medicoes)

ultima_medicao = medicoes.pop()
print("Valor removido:", ultima_medicao)
print("Lista restante:", medicoes)
```

## 11. Dois nomes podem apontar para a mesma lista

A atribuição `outro_nome = valores` não copia a lista. Os dois nomes passam a acessar o mesmo objeto. Uma alteração feita por um deles aparece quando consultamos o outro.

Para esta lista de números, `copy` cria outra lista que podemos alterar sem modificar a primeira. Essa cópia é superficial: se a lista contivesse outras listas, esses objetos internos continuariam compartilhados.

**Preveja:** qual lista ficará com `99.0`? Qual ficará com `-1.0`?


```python
valores = [1.0, 2.0, 3.0]
outro_nome = valores
copia_valores = valores.copy()

outro_nome[0] = 99.0
copia_valores[1] = -1.0

print("Lista original:", valores)
print("Outro nome:", outro_nome)
print("Cópia:", copia_valores)
```

## 12. Tuplas: agrupar valores

Uma tupla é uma sequência que não permite substituir, acrescentar ou remover seus itens depois de criada. Podemos usá-la para agrupar as coordenadas de um ponto ou os extremos de um intervalo.

O desempacotamento associa cada item a um nome. O número de nomes precisa corresponder ao número de itens.

**Altere:** mude as coordenadas na criação de `ponto` e execute a célula novamente. Isso cria uma nova tupla.


```python
ponto = (0.25, 0.75)
coordenada_x, coordenada_y = ponto

print(ponto)
print(type(ponto))
print(f"x = {coordenada_x}; y = {coordenada_y}")
print(f"Primeira coordenada: {ponto[0]}")
```

## 13. Dicionários: dados identificados por nomes

Um dicionário associa chaves a valores. Em vez de lembrar que uma propriedade ocupa certa posição, podemos consultá-la por uma chave, como `"condutividade"`.

Dicionários são mutáveis. `dicionario[chave]` exige que a chave exista; `get` permite fornecer um valor padrão para uma chave ausente.

**Altere:** acrescente uma chave `"temperatura"` e apresente seu valor.


```python
material = {
    "nome": "material de exemplo",
    "condutividade": 2.5,
    "unidade": "W/(m K)",
}

print(material["nome"])
print(material["condutividade"], material["unidade"])

material["condutividade"] = 3.0
print("Valor atualizado:", material["condutividade"])
print("Densidade:", material.get("densidade", "não informada"))
```

## 14. Conjuntos: valores distintos

Um conjunto (`set`) guarda itens distintos. Ele é útil para identificar categorias presentes nos dados ou testar pertencimento com `in`.

Conjuntos não são acessados por índices e não devemos depender de sua ordem de apresentação. Abaixo usamos `sorted` apenas para exibir os nomes em ordem. Um conjunto vazio é criado com `set()`; `{}` cria um dicionário vazio.

**Preveja:** a palavra repetida aumenta o tamanho do conjunto?


```python
regioes = {"interior", "fronteira", "interior"}

print(sorted(regioes))
print(f"Número de regiões distintas: {len(regioes)}")
print("fronteira" in regioes)
print("exterior" in regioes)
```

## 15. Uma observação sobre números de ponto flutuante

Computadores representam números reais com precisão finita. Muitos números decimais, como `0.1`, não têm uma representação binária finita exata.

A soma abaixo está muito próxima de `0.3`, mas a comparação com `==` é falsa. Não se trata de uma mudança na regra da adição: estamos observando o efeito da representação aproximada dos números.

No caderno 03 usaremos `math.isclose` para verificar resultados com tolerâncias explícitas. A escolha da tolerância depende do cálculo; exibir menos casas decimais não aumenta sua precisão.


```python
soma = 0.1 + 0.2

print(soma)
print(soma == 0.3)
print(f"Apresentação com uma casa decimal: {soma:.1f}")
print(f"Diferença em relação a 0.3: {soma - 0.3}")
```
