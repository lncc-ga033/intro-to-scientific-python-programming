## 1. De uma lista a um array

Um **ndarray** guarda dados de um mesmo tipo e organiza esses dados em eixos. Um vetor possui um eixo; uma matriz possui dois. Aqui, cada posição é um número real.


```python
posicoes = np.array([0.0, 0.25, 0.50, 0.75, 1.0])

print("Valores:", posicoes)
print("shape:", posicoes.shape)
print("Número de eixos:", posicoes.ndim)
print("Número de valores:", posicoes.size)
print("dtype:", posicoes.dtype)
```

**shape** é uma tupla com o comprimento de cada eixo. O vetor acima tem shape **(5,)**; uma matriz com uma linha teria shape **(1, 5)**. São dimensões diferentes, embora ambos contenham cinco números.

O **dtype** descreve o tipo dos valores. Use ponto flutuante quando precisar representar frações: atribuir uma fração a um array de inteiros pode truncá-la.


```python
inteiros = np.array([1, 2, 3], dtype=int)
reais = inteiros.astype(float)
reais[0] = 1.5

linha = posicoes.reshape(1, 5)
print("Inteiros:", inteiros, inteiros.dtype)
print("Reais:", reais, reais.dtype)
print("Vetor:", posicoes.shape, "| matriz com uma linha:", linha.shape)
```

Para criar sequências, **linspace** recebe os extremos e o número de pontos; **arange** recebe início, fim excluído e passo. Para dividir um intervalo real em um número conhecido de pontos, linspace deixa essa intenção mais clara.


```python
pontos = np.linspace(0.0, 1.0, 5)
indices = np.arange(5)
zeros = np.zeros(3)

print("Pontos:", pontos)
print("Índices:", indices)
print("Zeros:", zeros)
```

## 2. Índices, fatias e máscaras

Os índices começam em zero. Uma fatia **inicio:fim** inclui o início e exclui o fim. Uma máscara é um array booleano: seleciona as posições onde a condição é verdadeira.


```python
print("Primeiro ponto:", pontos[0])
print("Último ponto:", pontos[-1])
print("Índices 1, 2 e 3:", pontos[1:4])

mascara = pontos >= 0.5
print("Máscara:", mascara)
print("Pontos selecionados:", pontos[mascara])
```

Vamos organizar temperaturas em graus Celsius em uma matriz. **Cada linha é um dia; cada coluna é um sensor.** Essa convenção será importante para as médias.

Em dois eixos, escrevemos **matriz[linha, coluna]**. O símbolo **:** seleciona todas as posições de um eixo.


```python
temperaturas = np.array(
    [
        [18.0, 19.0, 20.0],
        [20.0, 22.0, 24.0],
        [19.0, 21.0, 23.0],
    ]
)

print("shape:", temperaturas.shape)
print("Segundo dia:", temperaturas[1, :])
print("Terceiro sensor:", temperaturas[:, 2])
print("Dia 2, sensor 3:", temperaturas[1, 2])
```

## 3. Operações elemento a elemento e produto matricial

Em arrays, **+**, **-**, **\*** e **\*\*** operam elemento a elemento, se as dimensões forem compatíveis. O símbolo **@** representa o produto matricial.

Uma conversão de temperaturas usa uma operação elemento a elemento. A aplicação de uma matriz a um vetor combina valores por meio de somas de produtos.


```python
kelvin = temperaturas + 273.15
quadrados = pontos**2

A = np.array([[2.0, 1.0], [0.0, 3.0]])
v = np.array([4.0, 5.0])

print("Primeiro dia em kelvin:", kelvin[0])
print("Quadrados:", quadrados)
print("v * v:", v * v)
print("A @ v:", A @ v)
print("v @ v, produto interno:", v @ v)
```

Se os dois operandos são matrizes com o mesmo shape, **A \* B** e **A @ B** podem ser permitidos, mas calculam objetos diferentes. Antes de operar, identifique a fórmula matemática desejada.


```python
B = np.array([[1.0, 2.0], [3.0, 4.0]])

print("Produto elemento a elemento:\n", A * B)
print("Produto matricial:\n", A @ B)
```

## 4. O que significa axis?

Uma redução percorre o eixo indicado e o elimina do resultado.

- **axis=0** percorre as linhas: retorna uma média por **sensor**.
- **axis=1** percorre as colunas: retorna uma média por **dia**.
- Sem axis, a média usa todos os valores.

A interpretação depende da organização da tabela.


```python
media_por_sensor = temperaturas.mean(axis=0)
media_por_dia = temperaturas.mean(axis=1)

print("Média por sensor:", media_por_sensor)
print("Média por dia:", media_por_dia)
print("Média de toda a tabela:", temperaturas.mean())
```

## 5. Broadcasting: repetir uma operação sem repetir os dados

Queremos retirar, de cada sensor, a sua média. A tabela tem shape **(3, 3)** e o vetor de médias tem shape **(3,)**. O NumPy alinha as dimensões a partir da direita e aplica esse vetor a cada linha.

Dimensões correspondentes são compatíveis quando são iguais ou quando uma delas vale 1. Desenhar os shapes ajuda a evitar operações que são aceitas, mas não representam o cálculo desejado.

A verificação **np.testing.assert_allclose** fica silenciosa quando os valores respeitam as tolerâncias e gera um erro se a comparação falhar.


```python
desvios = temperaturas - media_por_sensor

print("Shapes:", temperaturas.shape, media_por_sensor.shape, "->", desvios.shape)
print("Desvios:\n", desvios)
print("Média dos desvios por sensor:", desvios.mean(axis=0))
np.testing.assert_allclose(desvios.mean(axis=0), 0.0, atol=1e-12)
```

## 6. Nomes, views e cópias

Dar um segundo nome a um array não copia seus dados. Uma **fatia simples** normalmente é uma **view**: acessa parte dos mesmos dados.

**Alterar um elemento** e **reatribuir um nome** são ações diferentes. Alterar a view muda o original; reatribuir o nome da view a um novo array não altera o original.


```python
original = np.array([10.0, 20.0, 30.0, 40.0])
outro_nome = original
trecho = original[1:3]

trecho[0] = 99.0
print("Original após alterar a view:", original)
print("Mesmo objeto?", outro_nome is original)
print("Compartilham dados?", np.shares_memory(original, trecho))

trecho = np.array([-1.0, -2.0])
print("Original após reatribuir o nome trecho:", original)
```

Use **copy()** para modificar uma fatia sem alterar a origem. Seleções com máscara ou índices inteiros avançados retornam cópias; não trate toda seleção como uma view.

Aqui usamos arrays numéricos. Cópias de arrays com objetos Python exigem cuidados adicionais com os objetos referenciados.


```python
base = np.array([10.0, 20.0, 30.0, 40.0])
copia = base[1:3].copy()
copia[0] = -100.0

print("Base:", base)
print("Cópia modificada:", copia)
print("Compartilham dados?", np.shares_memory(base, copia))
```

## 7. Aleatoriedade reproduzível e vetorização

Um gerador com semente explícita permite repetir o exemplo no mesmo ambiente. Seu estado avança a cada sorteio; execute novamente a célula inteira para reiniciar a sequência.

Vamos calcular os mesmos quadrados com um laço Python e com uma operação do NumPy.


```python
rng = np.random.default_rng(2026)
amostra = rng.normal(loc=0.0, scale=1.0, size=6)

por_laco = np.empty_like(amostra)
for i in range(amostra.size):
    por_laco[i] = amostra[i] ** 2

por_numpy = amostra**2
print("Amostra:", np.round(amostra, 3))
print("Resultados equivalentes?", np.allclose(por_laco, por_numpy))
```

As duas versões visitam os $n$ valores: ambas têm custo $\Theta(n)$ neste cálculo. A operação vetorizada costuma ser mais rápida porque o trabalho repetitivo acontece em rotinas compiladas, com menor sobrecarga por elemento.

Isso não significa que os laços deixaram de existir nem que a complexidade mudou. Ganhos de tempo dependem dos dados e do ambiente; não precisamos de um benchmark longo para compreender essa distinção.


## 8. Resolver um sistema e conferir a resposta

Para $Au=b$, use **np.linalg.solve(A, b)**. Calcular explicitamente a inversa não é necessário para encontrar $u$.

No sistema abaixo, podemos verificar à mão que $u=(1,2)^T$. O **resíduo** $r=Au-b$ mede quanto a resposta calculada deixa de satisfazer as equações.


```python
A_sistema = np.array([[3.0, -1.0], [-1.0, 2.0]])
b_sistema = np.array([1.0, 3.0])

u_sistema = np.linalg.solve(A_sistema, b_sistema)
residuo = A_sistema @ u_sistema - b_sistema

print("Solução:", u_sistema)
print("Resíduo:", residuo)
print("Norma do resíduo:", np.linalg.norm(residuo))
np.testing.assert_allclose(u_sistema, [1.0, 2.0], atol=1e-12)
```
